# CIFAR-10 Classifiers and ResNet-18

A compact walkthrough of fully connected, convolutional, and residual classifiers in PyTorch. Reusable code lives in `src/cifar10_resnet`; this notebook focuses on the experiment story.

## Goal

Compare progressively stronger image-classification architectures and make the training path reproducible.

```python
from pathlib import Path
import pandas as pd
from cifar10_resnet.models import FCNN, SimpleCNN, ResNet18, TunedCNN, count_trainable_parameters

models = {
    'FCNN': FCNN([3 * 32 * 32, 100, 10]),
    'Simple CNN': SimpleCNN(),
    'ResNet-18': ResNet18(),
    'Tuned CNN': TunedCNN(),
}
{name: count_trainable_parameters(model) for name, model in models.items()}
```

## Architecture notes

The custom ResNet-18 uses two-layer residual blocks and projection shortcuts when spatial or channel dimensions change. The tuned CNN uses augmentation, batch normalization, dropout, label smoothing, and cosine learning-rate decay.
